**Introdução e Estratégia**

O objetivo é encontrar múltiplas raízes num intervalo $[a, b]$. O método da Bisseção ou Falsa Posição padrão só encontra uma raiz por vez.
Para resolver isso, implementei uma estratégia de Busca Global (Discretização):

1.  Divido o intervalo total em $N$ subintervalos menores.
2.  Verifico se há troca de sinal nas pontas de cada subintervalo ($f(x_i) \cdot f(x_{i+1}) < 0$).

1.   Se houver troca de sinal, chamo o método da Bisseção apenas para aquele pedacinho para refinar o valor da raiz.

2.   Armazeno as raízes encontradas e filtro valores duplicados.





**Implementação**

```python
# Importação das Bibliotecas usadas

import numpy as np
import matplotlib.pyplot as plt
```

```python
# Função do Método da Bisseção

def bissecao(f, a, b, tol=1e-6, max_iter=100):
    """
    Encontra uma raiz no intervalo [a, b] se houver troca de sinal
    """

    fa = f(a)
    fb = f(b)

    # Se não tiver troca de sinal, retorna None (segurança)
    if fa * fb > 0:
      return None

    iteracao = 0
    while (b - a) / 2 > tol and iteracao < max_iter:
      c = (a + b) / 2
      fc = f(c)

      if fc == 0:
        return c

      if fa * fc < 0:
        b = c
        fb = fc
      else:
        a = c
        fa = fc
      iteracao += 1

    return (a + b) / 2
```

```python
# Função Busca Global

def buscar_todas_raizes(f, a, b, n_divisoes=100):
  """
  Varre o intervalo [a, b] dividindo-o em n partes.
  Se detectar troca de sinal, chama a bisseção
  """

  x_pontos = np.linspace(a, b, n_divisoes + 1)
  raizes_encontradas = []

  print(f"Varrendo intervalo [{a}, {b} com {n_divisoes} subintervalos]")

  for i in range(len(x_pontos) - 1):
    x_esquerda = x_pontos[i]
    x_direita = x_pontos[i+1]

    # Checa troca de sinal (Teorema de Bolzano)
    if f(x_esquerda) * f(x_direita) <= 0:
      raiz = bissecao(f, x_esquerda, x_direita)

      if raiz is not None:
        # Evita salvar raízes repetidas
        # Só adiciona se a lista estiver vazia ou se for diferente da última
        if not raizes_encontradas or abs(raiz - raizes_encontradas[-1]) > 1e-4:
          raizes_encontradas.append(raiz)

    return raizes_encontradas
```

**Teste 1**

```python
# Definindo a função
f1 = lambda x: x**3 - 10*x**2 + 5

# Buscando raízes
intervalo_a, intervalo_b = -2, 11
raizes_f1 = buscar_todas_raizes(f1, intervalo_a, intervalo_b, n_divisoes=50)

print(f"\nRaízes encontradas para f1: {raizes_f1}")

# --- Gráfico para provar visualmente ---
x = np.linspace(intervalo_a, intervalo_b, 200)
y = f1(x)

plt.figure(figsize=(10, 5))
plt.plot(x, y, label='f(x) = x³ - 10x² + 5')
plt.axhline(0, color='black', linewidth=1) # Eixo X
plt.scatter(raizes_f1, [f1(r) for r in raizes_f1], color='red', zorder=5, label='Raízes Encontradas')
plt.title('Teste 1: Polinômio')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Definindo a função transcendental
f2 = lambda x: np.exp(-0.1*x) * np.sin(3*x)

# Buscando raízes
intervalo_a, intervalo_b = 0, 7 # Intervalo menor pra ver melhor
raizes_f2 = buscar_todas_raizes(f2, intervalo_a, intervalo_b, n_divisoes=100)

print(f"\nRaízes encontradas para f2: {raizes_f2}")

# Gráfico
x = np.linspace(intervalo_a, intervalo_b, 200)
y = f2(x)

plt.figure(figsize=(10, 5))
plt.plot(x, y, label='f(x) = e^(-0.1x) * sin(3x)')
plt.axhline(0, color='black', linewidth=1)
plt.scatter(raizes_f2, [f2(r) for r in raizes_f2], color='red', zorder=5, label='Raízes Encontradas')
plt.title('Teste 2: Função Transcendental Oscilatória')
plt.legend()
plt.grid(True)
plt.show()
```
